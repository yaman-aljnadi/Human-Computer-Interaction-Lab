```python
import sounddevice as sd
from scipy.io.wavfile import write
import numpy as np

# --- Configuration ---
# Sampling frequency (Hz)
freq = 44100
# Recording duration (seconds)
duration = 5

# Display recording information
print(f"Recording for {duration} seconds at {freq} Hz...")

# --- Recording process ---
# Start recorder with the given values of duration and sample frequency
# Records stereo audio (channels=2)
recording = sd.rec(int(duration * freq), samplerate=freq, channels=2)

# Record audio for the given number of seconds and wait for it to complete
sd.wait()

print("Recording complete. Saving file...")

# --- Saving the audio file ---
# This will convert the NumPy array to an audio file with the given sampling frequency
write("recording0.wav", freq, recording)

print("Audio saved as 'recording0.wav'")
```
